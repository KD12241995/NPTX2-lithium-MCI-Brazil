```python
import pandas as pd
from scipy.stats import pearsonr
import numpy as np

# 1. 상관관계 계산 함수 정의
def calculate_correlation(df1, df2, name1, name2):
    """
    두 DataFrame 간의 피어슨 상관계수와 p-value 계산
    """
    corr_results = []
    for col1 in df1.columns:
        for col2 in df2.columns:
            # 결측치 제거
            valid_idx = df1[col1].notna() & df2[col2].notna()
            if sum(valid_idx) > 3:  # 최소 3개 이상의 유효한 데이터 필요
                corr, pval = pearsonr(df1.loc[valid_idx, col1], df2.loc[valid_idx, col2])
                corr_results.append({
                    f'{name1}_Column': col1,
                    f'{name2}_Column': col2,
                    'Pearson_r': corr,
                    'p_value': pval
                })
    return pd.DataFrame(corr_results)

# 2. 모든 조합에 대해 상관관계 계산
results = []

# ELISA vs NULISA
if not elisa_data.empty and not nulisa_data.empty:
    df = calculate_correlation(elisa_data, nulisa_data, 'ELISA', 'NULISA')
    results.append(df)

# ELISA vs PRM
if not elisa_data.empty and not prm_data.empty:
    df = calculate_correlation(elisa_data, prm_data, 'ELISA', 'PRM')
    results.append(df)

# ELISA vs Demographic
if not elisa_data.empty and not demographic_data.empty:
    df = calculate_correlation(elisa_data, demographic_data, 'ELISA', 'Demographic')
    results.append(df)

# NULISA vs PRM
if not nulisa_data.empty and not prm_data.empty:
    df = calculate_correlation(nulisa_data, prm_data, 'NULISA', 'PRM')
    results.append(df)

# NULISA vs Demographic
if not nulisa_data.empty and not demographic_data.empty:
    df = calculate_correlation(nulisa_data, demographic_data, 'NULISA', 'Demographic')
    results.append(df)

# PRM vs Demographic
if not prm_data.empty and not demographic_data.empty:
    df = calculate_correlation(prm_data, demographic_data, 'PRM', 'Demographic')
    results.append(df)

# 3. 모든 결과 병합
final_results = pd.concat(results, ignore_index=True)

# 4. Excel 파일로 저장
output_path = "../data/output/correlation_results.xlsx"
final_results.to_excel(output_path, index=False)

print(f"상관관계 분석 결과가 '{output_path}'에 저장되었습니다.")
print("저장된 데이터 구조:")
print(final_results.head())
```

```python
import pandas as pd
from scipy.stats import pearsonr
import numpy as np

# 1. 상관관계 계산 함수 정의
def calculate_correlation(df1, df2, name1, name2):
    """
    두 DataFrame 간의 피어슨 상관계수와 p-value 계산
    """
    corr_results = []
    for col1 in df1.columns:
        for col2 in df2.columns:
            # 결측치 제거
            valid_idx = df1[col1].notna() & df2[col2].notna()
            if sum(valid_idx) > 3:  # 최소 3개 이상의 유효한 데이터 필요
                corr, pval = pearsonr(df1.loc[valid_idx, col1], df2.loc[valid_idx, col2])
                corr_results.append({
                    f'{name1}_Column': col1,
                    f'{name2}_Column': col2,
                    'Pearson_r': corr,
                    'p_value': pval
                })
    return pd.DataFrame(corr_results)

# 2. 각 조합별로 별도 파일 저장
def save_correlation_results(df1, df2, name1, name2, file_suffix):
    if not df1.empty and not df2.empty:
        result_df = calculate_correlation(df1, df2, name1, name2)
        if not result_df.empty:
            output_file = f"correlation_{file_suffix}.xlsx"
            result_df.to_excel(output_file, index=False)
            print(f"{name1} vs {name2} 상관관계 결과가 '{output_file}'에 저장되었습니다.")
            print("저장된 데이터 구조:")
            print(result_df.head())
        else:
            print(f"{name1} vs {name2}: 유효한 상관관계 결과가 없습니다.")
    else:
        print(f"{name1} 또는 {name2} 데이터가 비어있어 분석을 수행하지 않았습니다.")

# 3. 모든 조합에 대해 상관관계 계산 및 저장
# ELISA vs ELISA
save_correlation_results(elisa_data, elisa_data, 'ELISA', 'ELISA', 'elisa_vs_elisa')

# NULISA vs NULISA
save_correlation_results(nulisa_data, nulisa_data, 'NULISA', 'NULISA', 'nulisa_vs_nulisa')

# PRM vs PRM
save_correlation_results(prm_data, prm_data, 'PRM', 'PRM', 'prm_vs_prm')

print("모든 상관관계 분석이 완료되었습니다.")
```

```python
# 파일 경로 설정
file_path = '../data/input/Final_statistical_analysis.xlsx'

# 엑셀 파일 불러오기 (전체 데이터 사용)
data = pd.read_excel(file_path, sheet_name='Sheet1')
```

```python
# 칼럼 분류 함수 (수정된 버전)
def classify_columns(df):
    cols = {
        'demographic': ['CDR', 'CDR_SB', 'ADAS', 'SLN', 'TMTA', 'TMTB', 'Number of APOE Ɛ4'],
        'elisa': [col for col in df.columns if '_ELISA' in col.upper()],
        'nulisa': [col for col in df.columns if '_NULISA' in col.upper()],
        'prm': [col for col in df.columns if '_PRM' in col.upper()]
    }
    # 존재하는 칼럼만 선택 (괄호 수정)
    for k in cols:
        cols[k] = [col for col in cols[k] if col in df.columns]
    return cols

# 데이터 분리
cols = classify_columns(data)
demographic_data = data[cols['demographic']] if cols['demographic'] else pd.DataFrame()
elisa_data = data[cols['elisa']] if cols['elisa'] else pd.DataFrame()
nulisa_data = data[cols['nulisa']] if cols['nulisa'] else pd.DataFrame()
prm_data = data[cols['prm']] if cols['prm'] else pd.DataFrame()
```

```python
# 칼럼 분류 함수
def classify_columns(df):
    cols = {
        'demographic': ['CDR', 'CDR_SB', 'ADAS', 'SLN', 'TMTA', 'TMTB', 'Number of APOE Ɛ4'],
        'elisa': [col for col in df.columns if '_ELISA' in col.upper()],
        'nulisa': [col for col in df.columns if '_NULISA' in col.upper()],
        'prm': [col for col in df.columns if '_PRM' in col.upper()]
    }
    # 존재하는 칼럼만 선택
    for k in cols:
        cols[k] = [col for col in cols[k] if col in df.columns]
    return cols

# 데이터 분리
cols = classify_columns(data)
demographic_data = data[cols['demographic']] if cols['demographic'] else pd.DataFrame()
elisa_data = data[cols['elisa']] if cols['elisa'] else pd.DataFrame()
nulisa_data = data[cols['nulisa']] if cols['nulisa'] else pd.DataFrame()
prm_data = data[cols['prm']] if cols['prm'] else pd.DataFrame()

# 상관관계 계산 함수
def calculate_correlation(df1, df2, name1, name2):
    corr_results = []
    for col1 in df1.columns:
        for col2 in df2.columns:
            valid_idx = df1[col1].notna() & df2[col2].notna()
            if sum(valid_idx) > 3:
                corr, pval = pearsonr(df1.loc[valid_idx, col1], df2.loc[valid_idx, col2])
                corr_results.append({
                    'Comparison': f"{name1}_{col1}_vs_{name2}_{col2}",
                    'Group1': name1,
                    'Column1': col1,
                    'Group2': name2,
                    'Column2': col2,
                    'Pearson_r': corr,
                    'p_value': pval,
                    'N_valid_pairs': sum(valid_idx)
                })
    return pd.DataFrame(corr_results)

# 모든 가능한 조합 생성
combinations = [
    ('ELISA', 'NULISA', elisa_data, nulisa_data),
    ('ELISA', 'PRM', elisa_data, prm_data),
    ('ELISA', 'Demographic', elisa_data, demographic_data),
    ('NULISA', 'PRM', nulisa_data, prm_data),
    ('NULISA', 'Demographic', nulisa_data, demographic_data),
    ('PRM', 'Demographic', prm_data, demographic_data)
]

# 각 조합별로 분석 수행
all_results = []
for name1, name2, df1, df2 in combinations:
    if not df1.empty and not df2.empty:
        result = calculate_correlation(df1, df2, name1, name2)
        if not result.empty:
            output_file = f"correlation_{name1}_vs_{name2}.xlsx"
            result.to_excel(output_file, index=False)
            print(f"[{name1} vs {name2}] 결과 저장 완료: {output_file}")
            print(result.head(2), '\n')
            all_results.append(result)

# 전체 결과 통합
if all_results:
    final_results = pd.concat(all_results, ignore_index=True)
    final_results.to_excel("../data/output/correlation_all_combined.xlsx", index=False)
    print("모든 결과가 ../data/output/correlation_all_combined.xlsx에 통합 저장되었습니다.")
else:
    print("분석 가능한 유효한 데이터 조합이 없습니다.")

print("분석이 완료되었습니다.")
```
