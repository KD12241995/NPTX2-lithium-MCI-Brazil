# 04 – YWHAZ:NPTX2 Ratio vs. Cognitive Outcomes

Correlations between YWHAZ:NPTX2 ratio (ELISA/NULISA/PRM) and cognitive measures at baseline, 1-year, and change scores.

**Input:** `data/input/Final_statistical_analysis_YWHAZ_NPTX2_demographic.xlsx`, `data/input/Visit_Differences_YWHAZtoNPTX2.xlsx`

**Corresponds to:** Supplementary Figure S2

---

> **Running locally:** All paths are relative to the `notebooks/` folder.
> Input data files go in `data/input/`, outputs are written to `data/output/`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import mannwhitneyu
import warnings
```

```python
# 파일 경로 설정 (파일 경로는 본인의 실제 경로로 수정해야 합니다)
file_path = '../data/input/Final_statistical_analysis_YWHAZ_NPTX2_demographic.xlsx'
```

```python
# 엑셀 파일 읽기
df = pd.read_excel(file_path, sheet_name='Sheet1')
```

```python
# 1. 그룹 생성
# '_base'로 끝나는 셀들을 하나의 그룹으로 묶음
base_group = df[df['Group'].str.endswith('_base', na=False)].copy()
base_group['Group_category'] = 'Base'
```

```python
# 'Lithium_1y' 그룹
lithium_1y_group = df[df['Group'] == 'Lithium_1y'].copy()
lithium_1y_group['Group_category'] = 'Lithium_1y'
```

```python
# 'Placebo_1y' 그룹
placebo_1y_group = df[df['Group'] == 'Placebo_1y'].copy()
placebo_1y_group['Group_category'] = 'Placebo_1y'
```

```python
# 2. Demographic 정보 추출
demographic_columns = ['CDR', 'CDR_SB', 'ADAS', 'SLN', 'TMTA', 'TMTB', 'Number of APOE Ɛ4', 'Litemia']
```

```python
target_columns = ['YWHAZtoNPTX2_ELISA', 'YWHAZtoNPTX2_NULISA', 'YWHAZtoNPTX2_PRM']
```

```python
import pandas as pd
from scipy.stats import pearsonr
import numpy as np
```

```python
# 상관관계 분석 함수
def calculate_correlation(group, group_name):
    results = []
    for demo_col in demographic_columns:
        for target_col in target_columns:
            # 결측치 제거 후 분석
            valid_data = group[[demo_col, target_col]].dropna()
            if len(valid_data) >= 3:  # 최소 3개 이상의 데이터 포인트 필요
                r, p = pearsonr(valid_data[demo_col], valid_data[target_col])
                results.append({
                    'Group': group_name,
                    'Demographic': demo_col,
                    'Target': target_col,
                    'Pearson r': r,
                    'p-value': p
                })
    return pd.DataFrame(results)

# 각 그룹별 상관관계 계산
base_corr = calculate_correlation(base_group, 'Base')
lithium_corr = calculate_correlation(lithium_1y_group, 'Lithium_1y')
placebo_corr = calculate_correlation(placebo_1y_group, 'Placebo_1y')

# 결과 통합
all_results = pd.concat([base_corr, lithium_corr, placebo_corr], ignore_index=True)

# 엑셀 파일로 저장
output_path = '../data/output/YWHAZtoNPTX2_correlation_results.xlsx'
all_results.to_excel(output_path, index=False)

print("상관관계 분석 결과가 저장되었습니다:", output_path)
print(all_results.head())  # 결과 미리보기
```
